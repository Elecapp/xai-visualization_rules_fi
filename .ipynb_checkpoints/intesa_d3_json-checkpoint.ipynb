```python
def structure_json(json):
    new_json={}
    new_json['features']={}

    features = []
    # feature instances
    i=0
    for numerical_cat, eda in json['distribution']['numerical'].items():
        new_json['features'][i]['name']=numerical_cat
        new_json['features'][i]['rname']=numerical_cat
        new_json['features'][i]['eda']=eda
        new_json['features'][i]['type']='numeric'
        new_json['features'][i]['rule']={}
        new_json['features'][i]['crules']={}
        new_json['features'][i]['instance_value']=json['inst'][numerical_cat]
        
        i+=1
        features.append(numerical_cat)
    for categorical_cat, categorical_cat_dict in json['distribution']['categorical'].items():
        new_json['features'][i]['name']=categorical_cat
        new_json['features'][i]['rname']=categorical_cat
        new_json['features'][i]['eda']=None # todo
        new_json['features'][i]['type']='categorical'
        new_json['features'][i]['rule']={}
        new_json['features'][i]['crules']={}
        new_json['features'][i]['instance_value']=json['inst'][categorical_cat]
        
        
        i+=1
        features.append(categorical_cat)
    # feature importance
    for cat_index, cat_dict in new_json['features'].items():
        cat_name = cat_dict['name']
        new_json['features'][cat_index]['feature_importance']=json['feature_importance_all'][cat_name]
    # rules
    for old_rule_index, rule_dict in json['rule'].items():
        cat_name = rule_dict['att']
        pos_cat_name_in_features_list = features.index(cat_name)
        new_cat_rules = new_json[pos_cat_name_in_features_list]['rule']
        new_rule={}
        new_rule['att']= cat_name
        new_rule['op'] = rule_dict['op']
        new_rule['thr'] =  rule_dict['thr']
        new_rule['is_continuous']=rule_dict['is_continuous']
        new_json[pos_cat_name_in_features_list]['rule'][len(new_cat_rules)][0] = new_rule
        new_json[pos_cat_name_in_features_list]['rule'][len(new_cat_rules)][1] = 1

    # crules
        
        
        
        
    new_json['bb_pred']=1
    new_json['dd_pred']=1
    
    return new_json
```
